# FigS12c/d/e Min-system profiles

FigS12c is the individual-only unfolded `-MinC` profile panel; FigS12d is the corresponding `-MinDE` panel. FigS12e folds both filament halves and peak-normalizes each half-profile independently. The merged FigS12e output shows the mean with SE.

```python
import os
import sys
from pathlib import Path

import pandas as pd
from IPython.display import SVG, display
from matplotlib_inline.backend_inline import set_matplotlib_formats

os.environ.setdefault("OPENCV_LOG_LEVEL", "ERROR")
try:
    set_matplotlib_formats("svg")
except NotImplementedError:
    pass

try:
    from gear_data_analysis._bootstrap import bootstrap_notebook
except ModuleNotFoundError:
    def _find_repo_root(start: Path) -> Path:
        for candidate in (start, *start.parents):
            if (candidate / ".git").exists() or (
                (candidate / "pyproject.toml").exists()
                and (candidate / "src" / "gear_data_analysis" / "__init__.py").exists()
            ):
                return candidate
        raise FileNotFoundError("Repository root not found from current working directory.")

    REPO_ROOT = _find_repo_root(Path.cwd().resolve())
    SRC_ROOT = REPO_ROOT / "src"
    if str(SRC_ROOT) not in sys.path:
        sys.path.insert(0, str(SRC_ROOT))
    from gear_data_analysis._bootstrap import bootstrap_notebook

REPO_ROOT, SRC_ROOT = bootstrap_notebook(Path.cwd().resolve())
FIGURE_DIR = Path.cwd().resolve()


def preview_svgs(svg_paths):
    for svg_path in svg_paths:
        svg_path = Path(svg_path)
        if not svg_path.exists():
            raise FileNotFoundError(f"Missing SVG output: {svg_path}")
        display(SVG(filename=str(svg_path)))


from gear_data_analysis.peak_position_common import run_figs12cd_figure
from gear_data_analysis.split_profile_common import (
    build_figs12e_se_audit_table,
    run_figs12e_figure,
)

OUTPUT_DIR = FIGURE_DIR / "02_FigS12cde_plot_output"
OUTPUT_SVGS = [
    OUTPUT_DIR / "FigS12c.svg",
    OUTPUT_DIR / "FigS12d.svg",
    OUTPUT_DIR / "FigS12e.svg",
]
UNFOLDED_CSV = OUTPUT_DIR / "FigS12cd.csv"
MERGED_CSV = OUTPUT_DIR / "FigS12e.csv"
```

```python
unfolded_results = run_figs12cd_figure(FIGURE_DIR)
folded_results = run_figs12e_figure(FIGURE_DIR)

unfolded_summary = pd.DataFrame(
    [
        {
            "panel": result.source_key,
            "condition": result.plot_title,
            "n_filaments": result.frame_count,
        }
        for result in unfolded_results
    ]
)
folded_summary = pd.DataFrame(
    [
        {
            "condition": result.condition_label,
            "n_filaments": result.frame_count,
            "n_half_profiles": result.half_profile_count,
        }
        for result in folded_results
    ]
)
display(unfolded_summary)
folded_summary
```

```python
preview_svgs(OUTPUT_SVGS)
```

```python
unfolded_table = pd.read_csv(UNFOLDED_CSV)
merged_table = pd.read_csv(MERGED_CSV)

display(
    unfolded_table.loc[unfolded_table["table_type"] == "profile"]
    .groupby(["condition_key", "condition_label"], sort=False)
    .size()
    .rename("profile_bins")
    .to_frame()
)
merged_table.groupby(
    ["condition_key", "condition_label"], sort=False
).size().rename("folded_bins").to_frame()
```
