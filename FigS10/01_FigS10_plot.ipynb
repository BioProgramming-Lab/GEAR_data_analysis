# FigS10a–i – MinCDE oscillation-period analysis

## Figure contract

**Core conclusion.** Reconstituted mCherry-MinC undergoes pole-to-pole oscillation in GEARs, with strongly varying periods that show no detectable linear relationship with GEAR length.

- **FigS10a/e:** sample 01; 30 s interval; 57 frames; kymograph and FFT output.
- **FigS10b/f:** sample 02; 30 s interval; 74 frames; kymograph and FFT output.
- **FigS10c/g:** sample 04; 30 s interval; 62 frames; kymograph and FFT output.
- **FigS10d/h:** sample 06; 15 s interval; kymograph and FFT use the audited t001–t134 analysis window. Its additional c1/c2 frames t135–t149 remain in the raw folder but are not used.
- **FigS10i:** ordinary least-squares regression across ten independent GEARs. The four representative raw series are included here; the other six rows are retained as audited clean source data and are explicitly marked as not migrated.

This notebook exports quantitative results only: FigS10a–d are c2 kymographs, FigS10e–h are FFT spectra, and FigS10i is the length–period regression. It does not render microscopy frames, channel overlays, pole arrows or scale bars.

Plot appearance is inherited from the source workflows without restyling. Kymographs retain the original `Reds_Transparent` colormap, transparent SVG export, dimensions and colorbar; FFT and regression panels retain their original sizes, labels, line widths and marker settings, with only the four FFT panel titles omitted to match the submitted PDF.

Channel identities follow the final manuscript: c1 is the membrane geometry/annotation target, c2 is mCherry-MinC, and c3 is MreB-msfGFP. c3 is retained for source traceability and the audited common acquisition window but is not rendered.

For each sample, the c2 kymograph is normalized once by the global maximum of the complete analyzed series. The FFT trace is the intensity-weighted spatial center of mass across all c2 position bins, followed by linear detrending, mean subtraction, a Hann window and rFFT. The strongest non-zero bin with period ≥90 s is reported.
